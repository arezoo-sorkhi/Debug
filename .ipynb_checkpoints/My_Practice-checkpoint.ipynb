<p><strong>Task 1 : Count and Find prime numbers up to n</strong></p>

This function counts all prime numbers up to a given number n.

It also prints the list of primes found.

The inner loop checks divisibility only up to sqrt(num) for efficiency.


```python
# Count prime numbers up to n
def count_primes(n):
    count = 0  #keeps track of the total number of primes.
    primes = []  # list to store prime numbers

    for num in range(2, n + 1):
        is_prime = True # Assume the number is prime until we find a divisor

        for i in range(2, int(num**0.5) + 1): 
            #Using sqrt(num) optimization keeps it fast.
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            count += 1
            primes.append(num)  # add prime number to the list

    print("Prime numbers:", primes)
    return count


count_primes(9)
```

```python
# Find prime numbers up to n
def primes_up_to(n):
    return [num for num in range(2, n+1) if all(num % i != 0 for i in range(2, int(num**0.5)+1))]

print(primes_up_to(20))
```

```python
# Find prime numbers up to n
def primes_up_to(n):
    primes = []
    for num in range(2, n+1):
        is_prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                is_prime = False  # اگر تقسیم شد، عدد اول نیست
                break
        if is_prime:
            primes.append(num)
    return primes

print(primes_up_to(20))
```

<p><strong>Task 2 : QuickSort Note:</strong></p>


log(n) represents the number of times the list can be divided by 2
log(n) levels of recursion → O(n log n) average time complexity

Best & Average: O(n log n)

Worst: O(n²)

happens when pivot is always smallest or largest element,
e.g., in an already sorted list.

```python

# QuickSort example

def quicksort(arr):
    if len(arr) <= 1:
        return arr # Base case: a list of 0 or 1 element is already sorted
        
    pivot = arr[len(arr)//2] # Choose the middle element as pivot
    left = [x for x in arr if x < pivot]  # Elements smaller than pivot
    middle = [x for x in arr if x == pivot] # Elements equal to pivot
    right = [x for x in arr if x > pivot]  # Elements greater than pivot
    

    # Recursively sort left and right and combine
    return quicksort(left) + middle + quicksort(right)     



# Example usage
print(quicksort([3,6,8,10,1,2,1]))
```

<p><strong>Task 3 : Compute the mean and variance of a list of numbers and visualize them :
</strong></p>



```python
"""
NumPy & Matplotlib 
Compute the mean and variance of a list of numbers and visualize them.
"""

import numpy as np
import matplotlib.pyplot as plt

# 1️⃣ List of numbers
numbers = [2, 4, 6, 8]

# 2️⃣ Convert to NumPy array
arr = np.array(numbers)

# 3️⃣ Compute mean and variance
mean = np.mean(arr)
variance = np.var(arr)

# 4️⃣ Print results
print("Numbers:", numbers)
print("Mean:", mean)
print("Variance:", variance)


plt.figure(figsize=(6, 4))
plt.plot(arr, marker='o', linestyle='-', color='blue', label='Numbers')
plt.axhline(mean, color='red', linestyle='--', label=f'Mean = {mean}')
plt.title("Numbers and Their Mean")
plt.xlabel("Index")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```
